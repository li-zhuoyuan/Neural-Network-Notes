```python
x = torch.arange(4.0, requires_grad = True)
print(x)

y = 2 * torch.dot(x, x)
y
y.backward() ## 对 y进行反向传播，结果存在 x.grad中
print(x.grad)
x.grad == 4 * x

x.grad.zero_()
z = (x * x).sum()
z.backward()
print(x.grad)
```

```python
x.grad.zero_() # clear gradient
y = x * x
u = y.detach() # detach: 脱离，u把 y当作常数，而非关于 x的变量
z = u * x

z.sum().backward() # z是一个向量函数，需要先求和，在进行反向传播
x.grad == u
```

```python
def f(a):
    b = a * 2
    while b.norm() < 1000:
        b = b * 2
    if b.sum() > 0:
        c = b
    else:
        c = 100 * b
    return c

a = torch.randn(size=(), requires_grad=True)
d = f(a)
d.backward()

a.grad == d / a
```

```python
def try_all_gpus():  #@save
    """返回所有可用的GPU，如果没有GPU，则返回[cpu(),]"""
    devices = [torch.device(f'cuda:{i}')
             for i in range(torch.cuda.device_count())]
    return devices if devices else [torch.device('cpu')]
try_all_gpus()
```

```python
import torch
from torch import nn
# 创建一个包含 10 个单词，每个单词用 3 维向量表示的嵌入层
embedding = nn.Embedding(10, 3)
# 输入一个包含索引的张量
input_indices = torch.LongTensor([[1, 2, 4, 5], [4, 3, 2, 9]])
# 获取对应的嵌入向量
output = embedding(input_indices)
print(output)
```
